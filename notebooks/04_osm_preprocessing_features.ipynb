```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import os
import sys
sys.path.append(os.path.abspath("..")) 



df = pd.read_csv("../data/raw/osm_69_2025.csv", sep=',')
```

```python
# Vérifications
display(df)
print("\n",df.info())
print("total_rows:",df.shape[0],"total_columns:",df.shape[1])
print("\n",df.isna().sum())
print("\n","Duplicated Any:", df.duplicated().sum(),"\n")
print(df.info())

df.describe()

```

```python

# Colonnes NA, type variables, taux
documentation_variables = pd.DataFrame({
    "Valeurs manquantes": df.isnull().sum(),
    "Taux de NA (%)": (df.isna().mean() * 100).round(2).values,
    "Type informatique": df.dtypes.values})

print("\nDocumentation des variables:")
display(documentation_variables)

# Colonne distribution des valeurs
distribution = {}
liste = {}
for col in df.columns:
    if df[col].dtype in ["object", "string", "str"]:

        if df[col].nunique() <= 10:
            distribution[col] = ", ".join(df[col].dropna().astype(str).unique())
            liste[col]= "liste des catégories"
        else:
            distribution[col] = f"{df[col].nunique()} catégories"
            liste[col]= "nbr catégories"

    else:
        
        describe = df[col].describe()
        liste[col]= "Voir statistiques descriptives"
        distribution[col] = (
            f"Min={describe['min']:.2f} | "
            f"Médiane={describe['50%']:.2f} | "
            f"Moyenne={describe['mean']:.2f} | "
            f"Max={describe['max']:.2f}"
        )

documentation_distribution = pd.DataFrame({
    "Nom de la colonne": df.columns,
    "Type": df.dtypes.values,
    "Liste": [liste[col] for col in df.columns],
    "Distribution des valeurs": [distribution[col] for col in df.columns]
})

```

```python

#--------------------------------------------------------------
## PRE-PROCESSING  DES DONNÉES OSM
#--------------------------------------------------------------


osm = df.copy()


# Renommage des colonnes techniques
osm = osm.rename(columns={
    "@id": "osm_id",
    "@type": "osm_type",
    "@lat": "latitude",
    "@lon": "longitude",
    "@timestamp": "timestamp"
})


# Conversion des types : Timestamp OSM -> datetime
osm["timestamp"] = pd.to_datetime(osm["timestamp"],
    errors="coerce",
    # utc=True
)

# Coordonnées numériques
osm["latitude"] = pd.to_numeric(osm["latitude"], errors="coerce")
osm["longitude"] = pd.to_numeric(osm["longitude"], errors="coerce")

# ID OSM
osm["osm_id"] = pd.to_numeric(osm["osm_id"], errors="coerce")


# Suppression des colonnes peu utiles pour l'enrichissement
colonnes_a_supprimer = [
    "operator",
    "brand",
    "phone",
    "website",
    "opening_hours"
]

# name conservé pour les contrôles / analyses mais pas utilisé directement comme variable modèle
osm = osm.drop(columns=[c for c in colonnes_a_supprimer if c in osm.columns], errors="ignore")


# Ajout du département pour la traçabilité
osm["code_departement"] = "69"

## 5. Nettoyage des chaînes de caractères
# colonnes_texte = osm.select_dtypes(include="object").columns

# for col in colonnes_texte:
#     osm[col] = osm[col].astype("string").str.strip()


# Contrôle rapide des coordonnées
osm = osm[
    osm["latitude"].between(-90, 90, inclusive="both")
    & osm["longitude"].between(-180, 180, inclusive="both")
].copy()


# Suppression des doublons
osm = osm.drop_duplicates()

#---------------------------------------
## Contrôle final
#---------------------------------------

print("Shape finale :", osm.shape)
print("\nTypes :")
print(osm.dtypes)

print("\nValeurs manquantes :")
print(osm.isna().sum().sort_values(ascending=False).head(15))

print("\nTypes OSM :")
print(osm["osm_type"].value_counts(dropna=False))

osm.head()

osm.to_csv("../data/raw/osm_69_2025_preprocessed.csv",)

print("OSM terminé.")
```
